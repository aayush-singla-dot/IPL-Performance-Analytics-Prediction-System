```python
import pandas as pd

matches = pd.read_csv("../data/raw/matches.csv")
deliveries = pd.read_csv("../data/raw/deliveries.csv")

print(matches.shape)
print(deliveries.shape)
```

```python
matches.isnull().sum()
```

```python
matches = matches.dropna(subset=["winner"])

print(matches.shape)
```

```python
matches['city'] = matches['city'].fillna('Unknown')
```

```python
matches.replace({
    'Delhi Daredevils': 'Delhi Capitals',
    'Kings XI Punjab': 'Punjab Kings'
}, inplace=True)
```

```python
valid_teams = [
    'Chennai Super Kings',
    'Mumbai Indians',
    'Royal Challengers Bangalore',
    'Kolkata Knight Riders',
    'Sunrisers Hyderabad',
    'Rajasthan Royals',
    'Delhi Capitals',
    'Punjab Kings',
    'Lucknow Super Giants',
    'Gujarat Titans'
]

matches = matches[
    matches['team1'].isin(valid_teams) &
    matches['team2'].isin(valid_teams)
]
```

```python
print(matches.shape)
```

```python
print(matches['team1'].unique())
```

```python
matches['date'] = pd.to_datetime(matches['date'])
```

```python
matches = matches.sort_values(by="date")
```

```python
matches = matches.reset_index(drop=True)
```

```python
print(matches['team1'].unique())
print(matches.shape)
```

```python
matches.to_csv("../data/processed/matches_cleaned.csv", index=False)

deliveries.to_csv("../data/processed/deliveries_cleaned.csv", index=False)
```
